```python
import sys
import os

# Install required packages
!{sys.executable} -m pip install geopandas matplotlib requests

# Download India states GeoJSON
url = "https://raw.githubusercontent.com/india-in-data/india-states-2019/master/india_states.geojson"

import requests

response = requests.get(url)
response.raise_for_status()

with open("india_states.geojson", "wb") as f:
    f.write(response.content)

print("GeoJSON downloaded successfully!")
print("File:", os.path.abspath("india_states.geojson"))
```

```python
import geopandas as gpd
import matplotlib.pyplot as plt

india = gpd.read_file("india_states.geojson")

fig, ax = plt.subplots(figsize=(10, 12))

india.plot(
    ax=ax,
    edgecolor="black",
    linewidth=0.7
)

ax.set_title("India Map", fontsize=20)
ax.axis("off")

plt.show()
```
